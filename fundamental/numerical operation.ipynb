# Numerical Operations

```python
import math

a = 16
b = 4
```

Addition

```python
print(a + b)
```

Subtraction

```python
print(a - b)
```

Multiplication

```python
print(a * b)
```

Division

```python
print(a / b)
```

Floor Division

```python
print(a // b)
```

Modulus (Remainder)

```python
print(a % b)
```

Power

```python
print(a ** b)
```

Rounding

```python
print(round(3.14159, 2))
```

Built-in Math Functions

```python
print(math.sqrt(16))
print(math.pow(2, 3))
print(math.log(10))
print(math.exp(1))
```

Trigonometric Functions

```python
print(math.sin(math.pi / 2))
print(math.cos(0))
print(math.tan(math.pi / 4))
```

Constants

```python
print(math.pi)
print(math.e)
```
