# Floating Point Fundamentals

```python
import sys
import numpy as np
```

## Floating Point Representation

Computers cannot represent real numbers exactly. Instead, they use **floating point representation**, typically following the IEEE 754 standard. A floating point number is stored as:

* **Sign**
* **Exponent**
* **Mantissa (significand)**

In Python, we can inspect the properties of the default `float` (usually `float64`):

```python
sys.float_info
```

This reveals limits such as precision, minimum/maximum values, and machine epsilon.

## Decimal Numbers Are Not Exact in Binary

Some decimal numbers cannot be represented exactly in binary. A classic example is `0.1`:

```python
print(0.1)
print(format(0.1, ".24f"))
```

This happens because `0.1` is a repeating fraction in base-2, leading to a **representation error**.

## Machine Epsilon

Machine epsilon is the smallest number $( \epsilon )$ such that:

$$\epsilon_{machine} = \min{\epsilon > 0 \mid 1 + \epsilon \neq 1}$$

In Python:

```python
np.finfo(float).eps
```

This value defines the **precision limit** of floating point arithmetic.

## Precision Across Data Types

Different floating point types have different precision:

```python
for dtype in [np.float32, np.float64]:
    print(dtype, np.finfo(dtype).eps)
```

* `float32`: lower precision, faster, more memory-efficient
* `float64`: higher precision, standard for scientific computing

## Loss of Significance (Absorption)

Consider the following:

```python
a = 1e16
b = 1
print((a + b) - a)
```

Why?
* `a = 1e16` is very large
* `b = 1` is too small relative to `a`
* The floating point system cannot represent the small change

## Catastrophic Cancellation

Now consider:

```python
print(1 - np.cos(1e-8))
```

Why?
* $\cos(1e-8) \approx 1$ within machine precision
* The subtraction `1 - cos(x)` cancels leading digits
* The result falls **below machine epsilon**

This is known as **catastrophic cancellation**.

## Key Takeaways

* Floating point arithmetic is **approximate**, not exact
* Small numbers can be **lost** when added to large numbers
* Subtracting nearly equal numbers can cause **catastrophic cancellation**
* Machine epsilon defines the **precision limit**
* Numerical algorithms must be designed with **stability** in mind
