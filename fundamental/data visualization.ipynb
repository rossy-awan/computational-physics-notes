# Data Visualization

```python
import matplotlib.pyplot as plt
import numpy as np
```

## Basic Plot Configuration

We begin by configuring the global plotting style using `matplotlib`. This ensures consistency across all figures, which is especially important for scientific visualization.

```python
plt.rcParams.update({ "font.family": "Times New Roman", "mathtext.fontset": "cm", "font.size": 14 })
```

* **Font family** is set to *Times New Roman* for a publication-style appearance.
* **Math text font** uses Computer Modern (`cm`), matching standard LaTeX formatting.
* **Font size** is increased for better readability.

## Simple Line Plot

We first construct a basic function:

$$y = x^2$$

```python
x = np.linspace(0, 10, 101)
y = x**2
```

* `x` is a uniformly spaced array between 0 and 10.
* `y` is computed as a quadratic function of `x`.

```python
plt.figure(figsize=(4, 4))
plt.plot(x, y)
plt.xlabel(r"$x$")
plt.ylabel(r"$y$")
plt.tight_layout()
plt.show()
```

This produces a simple line plot:

* `plt.plot()` draws the curve.
* Axis labels are written in LaTeX format.
* `tight_layout()` ensures proper spacing.

## Multiple Functions in One Plot

We now visualize several functions simultaneously:

$$y_1 = x, \quad y_2 = x^2, \quad y_3 = x^3$$

```python
y1 = x
y2 = x**2
y3 = x**3
```

```python
plt.figure(figsize=(4, 4))
plt.plot(x, y1)
plt.plot(x, y2)
plt.plot(x, y3)
plt.legend([r"$x$", r"$x^2$", r"$x^3$"])
plt.xlabel(r"$x$")
plt.ylabel(r"$y$")
plt.tight_layout()
plt.show()
```

* Multiple datasets can be plotted on the same axes.
* A legend helps distinguish between curves.
* This is useful for comparing functional behavior.

## Subplots

To display multiple plots within the same figure, we use subplots:

```python
plt.figure(figsize=(8, 4))
plt.subplot(1, 2, 1)
plt.plot(x, x)
plt.title("Linear")
plt.subplot(1, 2, 2)
plt.plot(x, x**2)
plt.title("Quadratic")
plt.tight_layout()
plt.show()
```

* `subplot(1, 2, i)` creates a grid of 1 row and 2 columns.
* Each subplot can display a different function.
* Titles help clarify each panel.

## Scatter Plot

Scatter plots are useful for visualizing discrete data or random distributions.

```python
plt.figure(figsize=(5, 5))
plt.scatter(np.random.rand(100), np.random.rand(100))
plt.tight_layout()
plt.show()
```

* Random points are generated using uniform distribution.
* Each point is plotted independently.
* Useful in statistical physics or Monte Carlo simulations.

## Histogram

Histograms show the distribution of data values.

```python
plt.figure(figsize=(5, 5))
plt.hist(np.random.randn(10000), bins=30)
plt.tight_layout()
plt.show()
```

* Data is drawn from a normal (Gaussian) distribution.
* `bins=30` controls resolution of the histogram.
* Commonly used to analyze probability distributions.

## Error Bars

Experimental data often include uncertainties.

```python
y = x**2
noise = np.random.randn(len(x)) * 5
y_data = y + noise
y_err = np.full_like(x, 5)

plt.figure(figsize=(5, 5))
plt.errorbar(x, y_data, yerr=y_err, fmt="o", markersize=3, label="data")
plt.plot(x, y, label="model")
plt.legend()
plt.xlabel(r"$x$")
plt.ylabel(r"$y$")
plt.tight_layout()
plt.show()
```

* `errorbar` visualizes uncertainty.
* Important for comparing models with data.

## Heatmap (2D Visualization)

Two-dimensional data can be visualized using color maps.

```python
X, Y = np.meshgrid(x, x)
Z = np.sin(X) * np.cos(Y)

plt.figure(figsize=(6, 5))
plt.imshow(Z, extent=[0, 10, 0, 10], origin="lower", cmap="magma")
plt.colorbar(label=r"$z$")
plt.xlabel(r"$x$")
plt.ylabel(r"$y$")
plt.tight_layout()
plt.show()
```

* `imshow` maps values to colors.
* `colorbar` provides quantitative interpretation.

## Filled Contour Plot

A filled contour plot represents the scalar field using color regions.

```python
x = np.linspace(-5, 5, 300)
y = np.linspace(-5, 5, 300)
X, Y = np.meshgrid(x, y)
A1, x1, y1, s1 = 1.0, -1.5, 0.0, 1.0
A2, x2, y2, s2 = 1.0,  1.5, 0.0, 1.0
G1 = A1 * np.exp(-((X - x1)**2 + (Y - y1)**2) / (2 * s1**2))
G2 = A2 * np.exp(-((X - x2)**2 + (Y - y2)**2) / (2 * s2**2))
Z = G1 - G2
```

```python
plt.figure(figsize=(6, 5))
contour = plt.contourf(X, Y, Z, levels=10, cmap="Blues")
plt.colorbar(contour, label=r"$z$")
plt.xlabel(r"$x$")
plt.ylabel(r"$y$")
plt.tight_layout()
plt.show()
```

* `contourf` fills regions between contour levels.
* `coolwarm` is useful for positive/negative symmetry.

## Saving Figures

Figures can be exported for reports or publications.

```python
plt.figure(figsize=(4,4))
plt.plot(x, x**2)
plt.xlabel(r"$x$")
plt.ylabel(r"$y$")
plt.tight_layout()
plt.savefig("figure.svg", dpi=300)
```

* `dpi=300` ensures high resolution.
* Recommended for publication-quality output.
