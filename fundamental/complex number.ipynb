# Complex Numbers

```python
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams.update({ "font.family": "Times New Roman", "mathtext.fontset": "cm", "font.size": 14 })
```

## Definition of a Complex Number

A complex number is defined as:

$$z = x + iy$$

where:

* $x$ = real part
* $y$ = imaginary part
* $i = \sqrt{-1}$

Example in Python:

```python
z = 3 + 4j
print(z.real, z.imag)
```

## Modulus (Magnitude)

The magnitude (or modulus) of a complex number is:

$$r = |z| = \sqrt{x^2 + y^2}$$

Example:

```python
r = np.abs(z)
print(r)
```

## Argument (Phase Angle)

The phase (argument) is:

$$\theta = \arg(z) = \tan^{-1}\left(\frac{y}{x}\right)$$

In practice, we use `atan2` to get the correct quadrant:

```python
theta = np.angle(z)
print(theta)
```

## Polar Form

A complex number can be written in polar form:

$$z = r e^{i\theta}$$

This is known as the **exponential form**.

## Trigonometric Form

Using Euler’s formula:

$$e^{i\theta} = \cos\theta + i\sin\theta$$

So:

$$z = r(\cos\theta + i\sin\theta)$$

## Verification in Python

```python
z_reconstructed = r * (np.cos(theta) + 1j * np.sin(theta))
print("Original:", z)
print("Reconstructed:", z_reconstructed)
```

## Visualization on the Complex Plane

A complex number can be visualized as a point in the 2D plane:

* $x$-axis → real part
* $y$-axis → imaginary part

So:

$$z = x + iy \rightarrow (x, y)$$

```python
x = z.real
y = z.imag
t = np.linspace(0, theta, 100)

plt.figure()
plt.quiver(0, 0, x, y, angles="xy", scale_units="xy", scale=1)
plt.plot(np.cos(t), np.sin(t), "k")
plt.text(1.05*x, 1.05*y, r"$r$")
plt.text(1.1*np.cos(theta/2), 1.1*np.sin(theta/2), r"$\theta$")
plt.xlabel(r"$x$")
plt.ylabel(r"$y$")
plt.xlim(0, 5)
plt.ylim(0, 5)
plt.gca().set_aspect("equal", adjustable="box")
plt.tight_layout()
plt.show()
```

## Basic Arithmetic Operations

Complex Conjugate

$$z = x + iy \rightarrow \bar{z} = x - iy$$

```python
print(np.conj(z))
```

Let:

$$z_1 = x_1 + i y_1, \quad z_2 = x_2 + i y_2$$

```python
z1 = 2 + 3j
z2 = 1 + 4j
```

Addition

$$z_1 + z_2 = (x_1 + x_2) + i(y_1 + y_2)$$

```python
print(z1 + z2)
```

Subtraction

$$z_1 - z_2 = (x_1 - x_2) + i(y_1 - y_2)$$

```python
print(z1 - z2)
```

Multiplication

$$z_1 z_2 = (x_1x_2 - y_1y_2) + i(x_1y_2 + x_2y_1)$$

```python
print(z1 * z2)
```

Division:

$$\frac{z_1}{z_2} = \frac{z_1 \cdot \bar{z}_2}{|z_2|^2}$$

```python
print(z1 / z2)
```

Powers (De Moivre’s Theorem)

$$z^n = r^n e^{i n\theta}$$
