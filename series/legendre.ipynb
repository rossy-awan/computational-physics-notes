# Legendre Polynomials

Legendre polynomials are a sequence of orthogonal polynomials defined on the interval $[-1,1]$. They are denoted by $P_n(x)$, where $n$ is the polynomial degree. The first few Legendre polynomials are

$$
P_0(x) = 1,\quad
P_1(x) = x,\quad
P_2(x) = \frac{1}{2}(3x^2-1),\quad
P_3(x) = \frac{1}{2}(5x^3-3x).
$$

Unlike an ordinary polynomial expansion in powers of $x$, a function can be represented using the Legendre basis as

$$
f(x) = \sum_{n=0}^{\infty}a_nP_n(x).
$$

The Legendre polynomials are orthogonal on $[-1,1]$:

$$
\int_{-1}^{1}P_m(x)P_n(x)\,dx = \frac{2}{2n+1}\delta_{mn},
$$

where $\delta_{mn}$ is the Kronecker delta.

```python
import numpy as np
import matplotlib.pyplot as plt
from numpy.polynomial import Legendre
```

```python
plt.rcParams.update({ "font.family": "Times New Roman", "mathtext.fontset": "cm", "font.size": 14 })
```

## Legendre Expansion

Consider the function

$$
V(\theta) = x^2,\quad
x = \cos\theta.
$$

Since $\theta\in[0,\pi]$, the corresponding variable satisfies $x\in[-1,1]$, making Legendre polynomials a natural basis for this representation.

```python
theta = np.linspace(0, np.pi, 1000)
x = np.cos(theta)
V = x**2
deg = 2
```

A degree-$2$ Legendre approximation is obtained using `Legendre.fit`:

```python
P = Legendre.fit(x, V, deg)
P
```

The output shows the coefficients of the Legendre expansion. Thus,

$$
V(x)\approx\frac{1}{3}P_0(x) + \frac{2}{3}P_2(x).
$$

Using

$$
P_0(x) = 1,\quad
P_2(x) = \frac{1}{2}(3x^2-1),
$$

gives

$$
\frac{1}{3}P_0(x) + \frac{2}{3}P_2(x) = x^2.
$$

Therefore, the degree-$2$ Legendre representation reproduces the function exactly.

## Verification

The numerical approximation can be verified using `np.allclose`:

```python
np.allclose(P(x), V)
```

A result of `True` indicates that the values computed from the Legendre representation agree with the original function within the numerical tolerance.

## Approximation

Finally, the original function and its Legendre representation are plotted as functions of $\theta$:

```python
plt.figure(figsize=(5, 5))
plt.plot(theta, V, "k", label=r"$V(\theta)$")
plt.plot(theta, P(x), "r", label="Legendre")
plt.xlabel(r"$\theta$")
plt.ylabel(r"$V$")
plt.legend()
plt.tight_layout()
plt.show()
```

Since the two functions overlap, the plot provides a visual confirmation of the numerical result.
