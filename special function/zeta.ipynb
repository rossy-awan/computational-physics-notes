# Zeta Functions

The Riemann zeta function is a special function defined by the infinite series

$$
\zeta(s) = \sum_{n=1}^{\infty}\frac{1}{n^s},
$$

for $\operatorname{Re}(s)>1$. It can be extended to other values of $s$ through analytic continuation, except for a simple pole at

$$
s=1.
$$

A generalization of the Riemann zeta function is the Hurwitz zeta function,

$$
\zeta(s,q) = \sum_{n=0}^{\infty} \frac{1}{(n+q)^s},
$$

where $q>0$. The Riemann zeta function is recovered from the Hurwitz zeta function when

$$
q = 1,
$$

since

$$
\zeta(s,1) = \zeta(s).
$$

SciPy provides both forms through `special.zeta`.

```python
from scipy import special
import numpy as np
import matplotlib.pyplot as plt
```

```python
plt.rcParams.update({ "font.family": "Times New Roman", "mathtext.fontset": "cm", "font.size": 14 })
```

```python
def plot_zeta(type="riemann", q=1, s_min=0, s_max=10, num=1000):
    s = np.linspace(s_min, s_max, num)
    if type == "riemann":
        vals = special.zeta(s)
    elif type == "hurwitz":
        vals = special.zeta(s, q)
    plt.figure(figsize=(5, 5))
    plt.plot(s, vals, "r")
    plt.xlabel(r"$s$")
    plt.ylabel(r"$\zeta(s)$")
    plt.tight_layout()
    plt.show()
```

## Riemann Zeta Function

The Riemann zeta function can be evaluated using `special.zeta(s)`. For example,

```python
plot_zeta(type="riemann")
```

produces

$$
\zeta(s).
$$

The series representation shows that the function is closely related to sums over integer powers,

$$
\zeta(s) = 1 + \frac{1}{2^s} + \frac{1}{3^s} + \cdots.
$$

For real $s>1$, the series converges and $\zeta(s)$ decreases toward $1$ as $s$ increases. The function has a singularity at

$$
s = 1,
$$

where the series diverges.

## Hurwitz Zeta Function

The Hurwitz zeta function introduces an additional parameter $q$:

$$
\zeta(s,q) = \sum_{n=0}^{\infty} \frac{1}{(n+q)^s}.
$$

It can be evaluated using `special.zeta(s, q)`. For example,

```python
plot_zeta(type="hurwitz", q=0.5)
```

gives

$$
\zeta(s,0.5).
$$

Changing $q$ shifts the starting point of the series and therefore changes the value and behavior of the function. For $q=1$,

$$
\zeta(s,1) = \zeta(s),
$$

so the Riemann zeta function is a special case of the Hurwitz zeta function.
