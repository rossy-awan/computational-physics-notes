# Error Functions

The error function is a special function defined by the integral

$$
\operatorname{erf}(x) = \frac{2}{\sqrt{\pi}} \int_0^x e^{-t^2}\,dt.
$$

It is closely related to the Gaussian function

$$
e^{-x^2},
$$

and is therefore commonly encountered when evaluating integrals involving Gaussian distributions. The complementary error function is defined as

$$
\operatorname{erfc}(x) = 1 - \operatorname{erf}(x).
$$

SciPy provides both functions through `special.erf` and `special.erfc`.

```python
from scipy import special
import numpy as np
import matplotlib.pyplot as plt
```

```python
plt.rcParams.update({ "font.family": "Times New Roman", "mathtext.fontset": "cm", "font.size": 14 })
```

```python
def plot_erf(type="erf", x_min=-3, x_max=3, num=1000):
    x = np.linspace(x_min, x_max, num)
    if type == "erf":
        vals = special.erf(x)
        ylabel = r"$\operatorname{erf}(x)$"
    elif type == "erfc":
        vals = special.erfc(x)
        ylabel = r"$\operatorname{erfc}(x)$"
    else:
        raise ValueError("Type is not recognized.")
    plt.figure(figsize=(5, 5))
    plt.plot(x, vals, "r")
    plt.xlabel(r"$x$")
    plt.ylabel(ylabel)
    plt.tight_layout()
    plt.show()
```

## Error Function

The error function can be evaluated using `special.erf(x)`. For example,

```python
plot_erf(type="erf")
```

produces the function

$$
\operatorname{erf}(x).
$$

The function is odd,

$$
\operatorname{erf}(-x) = -\operatorname{erf}(x),
$$

and approaches $\pm1$ for large positive and negative values of $x$:

$$
\lim_{x\to\infty}\operatorname{erf}(x) = 1,\quad
\lim_{x\to-\infty}\operatorname{erf}(x) = -1.
$$

At the origin,

$$
\operatorname{erf}(0) = 0.
$$

## Complementary Error Function

The complementary error function is evaluated using `special.erfc(x)` and can be visualized with

```python
plot_erf(type="erfc")
```

Since

$$
\operatorname{erfc}(x) = 1 - \operatorname{erf}(x),
$$

the two functions satisfy

$$
\operatorname{erf}(x) + \operatorname{erfc}(x) = 1.
$$

In particular,

$$
\operatorname{erfc}(0) = 1,
$$

and

$$
\lim_{x\to\infty}\operatorname{erfc}(x) = 0.
$$
