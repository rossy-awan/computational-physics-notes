# Bessel Functions

Bessel functions arise as solutions of Bessel's differential equation,

$$
x^2\frac{d^2y}{dx^2} + x\frac{dy}{dx} + (x^2-\nu^2)y = 0,
$$

where $\nu$ is the order of the function. The two linearly independent solutions are the Bessel functions of the first and second kind,

$$
J_\nu(x),\quad
Y_\nu(x).
$$

For integer orders $n$, the functions are commonly written as

$$
J_n(x),\quad
Y_n(x).
$$

SciPy provides Bessel functions for both integer and arbitrary orders. The integer-order functions can be evaluated using `special.jn` and `special.yn`, while `special.jv` and `special.yv` allow the order to be a general real number.

```python
from scipy import special
import numpy as np
import matplotlib.pyplot as plt
```

```python
plt.rcParams.update({ "font.family": "Times New Roman", "mathtext.fontset": "cm", "font.size": 14 })
```

```python
def plot_bessel(family="J", orders=[0, 1, 2], x_min=0, x_max=20, num=1000):
    x = np.linspace(x_min, x_max, num)
    plt.figure(figsize=(5, 5))
    if family == "J":
        func = special.jn
        label = r"$J_{n}(x)$"
    elif family == "Jv":
        func = special.jv
        label = r"$J_{n}(x)$"
    elif family == "Y":
        func = special.yn
        label = r"$Y_{n}(x)$"
    elif family == "Yv":
        func = special.yv
        label = r"$Y_{n}(x)$"
    else:
        raise ValueError(f"Family '{family}' is not recognized.")
    for n in orders:
        plt.plot(x, func(n, x), label=label.replace("n", str(n)))
    plt.xlabel(r"$x$")
    plt.ylabel(r"$y$")
    plt.legend()
    plt.tight_layout()
    plt.show()
```

## Bessel Functions of the First Kind

The Bessel function of the first kind, $J_\nu(x)$, is regular at $x=0$ for non-negative orders. Integer-order functions can be evaluated using `special.jn`. For example,

```python
plot_bessel("J", orders=[0, 1, 2])
```

shows

$$
J_0(x),\quad
J_1(x),\quad
J_2(x).
$$

The order determines the shape of the oscillatory solution. Increasing the order changes both the behavior near the origin and the positions of the zeros.

The same family can be evaluated for non-integer orders using `special.jv`:

```python
plot_bessel("Jv", orders=[0.5, 1.5, 2.5])
```

This demonstrates that the order $\nu$ does not need to be an integer.

## Bessel Functions of the Second Kind

The Bessel function of the second kind, $Y_\nu(x)$, provides the second linearly independent solution of Bessel's equation. For integer orders, it can be evaluated using `special.yn`. For example,

```python
plot_bessel("Y", orders=[0, 1, 2], x_min=1)
```

gives

$$
Y_0(x),\quad
Y_1(x),\quad
Y_2(x).
$$

Unlike $J_\nu(x)$, the function $Y_\nu(x)$ is singular at the origin. The plot therefore starts from $x=1$ rather than $x=0$.

The arbitrary-order version is provided by `special.yv`:

```python
plot_bessel("Yv", orders=[0.5, 1.5, 2.5], x_min=1)
```

Thus, `jn` and `yn` are convenient for integer orders, while `jv` and `yv` provide the more general $\nu$-order functions.
