# Elliptic Functions

Jacobi elliptic functions are a family of special functions that generalize trigonometric functions to nonlinear problems. They are parameterized by the elliptic parameter $m$ and are commonly written as

$$
\operatorname{sn}(u\mid m),\quad
\operatorname{cn}(u\mid m),\quad
\operatorname{dn}(u\mid m).
$$

They are related to the Jacobi amplitude $\phi$ through

$$
u = F(\phi\mid m),
$$

where $F(\phi\mid m)$ is the incomplete elliptic integral of the first kind.

The Jacobi elliptic functions can be defined through

$$
\operatorname{sn}(u\mid m) = \sin\phi,
$$

$$
\operatorname{cn}(u\mid m) = \cos\phi,
$$

and

$$
\operatorname{dn}(u\mid m) = \sqrt{1-m\sin^2\phi}.
$$

SciPy provides these functions through `sn, cn, dn, ph = special.ellipj(u, m)` where `ph` is the Jacobi amplitude $\phi$.

```python
from scipy import special
import numpy as np
import matplotlib.pyplot as plt
```

```python
plt.rcParams.update({ "font.family": "Times New Roman", "mathtext.fontset": "cm", "font.size": 14 })
```

```python
def plot_jacobi(m=0.5, u_min=0, u_max=10, num=1000):
    u = np.linspace(u_min, u_max, num)
    sn, cn, dn, ph = special.ellipj(u, m)
    plt.figure(figsize=(5, 5))
    plt.plot(u, sn, "b", label=r"$\operatorname{sn}(u \mid m)$")
    plt.plot(u, cn, "r", label=r"$\operatorname{cn}(u \mid m)$")
    plt.plot(u, dn, "g", label=r"$\operatorname{dn}(u \mid m)$")
    plt.plot(u, ph, "k", label=r"$\phi(u \mid m)$")
    plt.xlabel(r"$u$")
    plt.legend()
    plt.tight_layout()
    plt.show()
```

## Jacobi Elliptic Functions

The functions can be visualized for different values of the parameter $m$:

```python
plot_jacobi(m=0)
```

For $m=0$, the Jacobi elliptic functions reduce to ordinary trigonometric functions:

$$
\operatorname{sn}(u\mid0) = \sin u,
$$

$$
\operatorname{cn}(u\mid0) = \cos u,
$$

and

$$
\operatorname{dn}(u\mid0) = 1.
$$

The Jacobi amplitude also becomes

$$
\phi(u\mid0) = u.
$$

Thus, the elliptic functions recover familiar trigonometric behavior in the limit $m=0$.

For an intermediate value such as $m=0.5$,

```python
plot_jacobi(m=0.5)
```

the functions no longer behave exactly like sine and cosine. The parameter $m$ controls the deviation from the ordinary trigonometric functions.

As $m$ approaches $1$, the behavior becomes increasingly non-periodic:

```python
plot_jacobi(m=1)
```

At $m=1$,

$$
\operatorname{sn}(u\mid1) = \tanh u,
$$

$$
\operatorname{cn}(u\mid1) = \operatorname{sech}u,
$$

and

$$
\operatorname{dn}(u\mid1) = \operatorname{sech}u.
$$

Thus, the limits

$$
m = 0
\quad\longrightarrow\quad
\text{trigonometric functions},
$$

and

$$
m = 1
\quad\longrightarrow\quad
\text{hyperbolic functions}
$$

show how Jacobi elliptic functions connect these two classes of functions.
