# Airy Functions

The Airy functions are special functions that arise as solutions of the differential equation

$$
\frac{d^2y}{dx^2} - xy = 0.
$$

This is known as the Airy equation. Its two linearly independent solutions are the Airy functions of the first and second kind,

$$
\operatorname{Ai}(x),\quad
\operatorname{Bi}(x).
$$

They satisfy

$$
\frac{d^2}{dx^2}\operatorname{Ai}(x) - x\operatorname{Ai}(x) = 0,
$$

and

$$
\frac{d^2}{dx^2}\operatorname{Bi}(x) - x\operatorname{Bi}(x) = 0.
$$

The derivatives of these functions are denoted by

$$
\operatorname{Ai}'(x) = \frac{d\operatorname{Ai}}{dx},\quad
\operatorname{Bi}'(x) = \frac{d\operatorname{Bi}}{dx}.
$$

SciPy provides both Airy functions and their derivatives through `scipy.special.airy`. The output contains

$$
\operatorname{Ai}(x),\quad
\operatorname{Ai}'(x),\quad
\operatorname{Bi}(x),\quad
\operatorname{Bi}'(x).
$$

```python
from scipy import special
import numpy as np
import matplotlib.pyplot as plt
```

```python
plt.rcParams.update({ "font.family": "Times New Roman", "mathtext.fontset": "cm", "font.size": 14 })
```

For example, the four quantities can be visualized together:

```python
def plot_airy(x_min=-10, x_max=10, num=1000):
    x = np.linspace(x_min, x_max, num)
    Ai, Aip, Bi, Bip = special.airy(x)
    plt.figure(figsize=(5, 5))
    plt.plot(x, Ai, "k", label=r"$A_i$")
    plt.plot(x, Aip, "k--", label=r"$\frac{dA_i}{dx}$")
    plt.plot(x, Bi, "r", label=r"$B_i$")
    plt.plot(x, Bip, "r--", label=r"$\frac{dB_i}{dx}$")
    plt.xlabel(r"$x$")
    plt.ylabel(r"$y$")
    plt.legend()
    plt.tight_layout()
    plt.show()
```

```python
plot_airy(x_min=-6, x_max=2, num=1000)
```

The Airy functions have different behavior for positive and negative values of $x$. In particular, $\operatorname{Ai}(x)$ decays for large positive $x$, while $\operatorname{Bi}(x)$ grows rapidly. For negative $x$, both functions exhibit oscillatory behavior.
