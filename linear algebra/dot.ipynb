# Dot Product

The dot product is a scalar operation between two vectors that produces a single numerical value. For vectors $\mathbf{a}$ and $\mathbf{b}$, the dot product is defined as

$$
\mathbf{a}\cdot\mathbf{b}=
\sum_{i=1}^{n} a_i b_i.
$$

For two vectors in three-dimensional space,

$$
\mathbf{a}\cdot\mathbf{b}=
a_x b_x + a_y b_y + a_z b_z.
$$

Using NumPy, the dot product can be computed with `np.dot`.

```python
import numpy as np
```

## Dot Product of Two Vectors

Consider the vectors

$$
\mathbf{a}=
\begin{bmatrix}
1\\
2\\
3
\end{bmatrix},
\quad
\mathbf{b}=
\begin{bmatrix}
4\\
5\\
6
\end{bmatrix}.
$$

The dot product combines the corresponding components of the two vectors and produces a scalar.

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
```

```python
np.dot(a, b)
```

The matrix multiplication operator `@` provides an equivalent notation for the dot product of one-dimensional NumPy arrays.

```python
a @ b
```

## Relation to the Angle Between Vectors

The dot product is related to the angle $\theta$ between two vectors by

$$
\mathbf{a}\cdot\mathbf{b}=\left\lVert\mathbf{a}\right\rVert\left\lVert\mathbf{b}\right\rVert\cos\theta.
$$

Therefore, the cosine of the angle between two nonzero vectors can be obtained from

$$
\cos\theta=\frac{\mathbf{a}\cdot\mathbf{b}}{\left\lVert\mathbf{a}\right\rVert\left\lVert\mathbf{b}\right\rVert}.
$$

This provides a direct connection between the algebraic dot product and the geometric angle between vectors.

```python
np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
```

## Orthogonal Vectors

Two vectors are orthogonal when their dot product is zero. Consider

$$
\mathbf{a}=
\begin{bmatrix}
1\\
0
\end{bmatrix},
\quad
\mathbf{b}=
\begin{bmatrix}
0\\
1
\end{bmatrix}.
$$

Their dot product is

$$
\mathbf{a}\cdot\mathbf{b}=0,
$$

confirming that the two vectors are perpendicular.

```python
a = np.array([1, 0])
b = np.array([0, 1])
```

```python
np.dot(a, b)
```

## Dot Product of Matrices

The dot product operation can also be applied to two-dimensional arrays. Consider

$$
A=
\begin{bmatrix}
1 & 2\\
3 & 4
\end{bmatrix},
\quad
B=
\begin{bmatrix}
5 & 6\\
7 & 8
\end{bmatrix}.
$$

For two-dimensional arrays, `np.dot(A, B)` performs matrix multiplication. The resulting matrix is obtained by taking the dot product of each row of $A$ with each column of $B$:

$$
AB=
\begin{bmatrix}
1(5)+2(7) & 1(6)+2(8)\\
3(5)+4(7) & 3(6)+4(8)
\end{bmatrix}.
$$

Thus, while `np.dot` computes a vector dot product for one-dimensional arrays, its behavior for two-dimensional arrays corresponds to matrix multiplication.

```python
A = np.array([[1, 2],
              [3, 4]])
B = np.array([[5, 6],
              [7, 8]])
```

```python
np.dot(A, B)
```
