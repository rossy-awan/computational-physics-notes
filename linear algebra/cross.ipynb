# Cross Product

The cross product is a vector operation that takes two vectors in three-dimensional space and produces a vector perpendicular to both. For vectors $\mathbf{a}$ and $\mathbf{b}$, the cross product is defined as

$$
\mathbf{a}\times\mathbf{b} =
\begin{vmatrix}
\mathbf{x} & \mathbf{y} & \mathbf{z}\\
a_x & a_y & a_z\\
b_x & b_y & b_z
\end{vmatrix}.
$$

The magnitude of the resulting vector is related to the angle $\theta$ between the two input vectors by

$$
\left\lVert\mathbf{a}\times\mathbf{b}\right\rVert =
\left\lVert\mathbf{a}\right\rVert
\left\lVert\mathbf{b}\right\rVert
\sin\theta\,\mathbf{n}.
$$

Using NumPy, the cross product is computed with `np.cross`.

```python
import numpy as np
```

## Cross Product of Two Vectors

Consider the vectors

$$
\mathbf{a}=
\begin{bmatrix}
1\\
2\\
3
\end{bmatrix},
\quad
\mathbf{b}=
\begin{bmatrix}
4\\
5\\
6
\end{bmatrix}.
$$

The cross product produces a vector $\mathbf{c}$ that is perpendicular to both $\mathbf{a}$ and $\mathbf{b}$.

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
c = np.cross(a, b)
c
```

The magnitude of $\mathbf{c}$ can then be evaluated using the Euclidean norm.

```python
np.linalg.norm(c)
```

## Orthogonality

A fundamental property of the cross product is that the resulting vector is orthogonal to both input vectors. This can be verified using the dot product:

$$
\mathbf{a}\cdot(\mathbf{a}\times\mathbf{b})=0,
\quad
\mathbf{b}\cdot(\mathbf{a}\times\mathbf{b})=0.
$$

Thus, the cross product provides a direct way to construct a vector normal to the plane spanned by $\mathbf{a}$ and $\mathbf{b}$.

```python
np.dot(a, c)
```

```python
np.dot(b, c)
```

## Relation to the Angle Between Vectors

The angle between two vectors can be related to the magnitude of their cross product through

$$
\sin\theta=
\frac{\left\lVert\mathbf{a}\times\mathbf{b}\right\rVert}{\left\lVert\mathbf{a}\right\rVert\left\lVert\mathbf{b}\right\rVert}.
$$

For the given vectors, this quantity can be evaluated directly and compared with the geometric interpretation of the cross product.


```python
np.linalg.norm(np.cross(a, b)) / (np.linalg.norm(a) * np.linalg.norm(b))
```

## Cross Product of Arrays

The cross product can also be applied to arrays containing multiple vectors. Consider

$$
A=
\begin{bmatrix}
1 & 2 & 3\\
0 & 1 & 0
\end{bmatrix},
\quad
B=
\begin{bmatrix}
4 & 5 & 6\\
1 & 0 & 0
\end{bmatrix}.
$$

Each corresponding pair of rows defines a pair of three-dimensional vectors. NumPy therefore computes the cross product for each pair, producing an array whose rows contain the resulting vectors.

```python
A = np.array([[1, 2, 3],
              [0, 1, 0]])
B = np.array([[4, 5, 6],
              [1, 0, 0]])
```

```python
np.cross(A, B)
```
