# Inner Product

The inner product generalizes the dot product to complex-valued vectors. For two complex vectors $\boldsymbol{\psi}$ and $\boldsymbol{\phi}$, the inner product is defined as

$$
\langle\phi|\psi\rangle=\sum_{i=1}^{n}\phi_i^*\psi_i,
$$

where $(\cdot)^*$ denotes complex conjugation. In NumPy, the conjugate is automatically applied to the first argument of `np.vdot`.

```python
import numpy as np
```

Consider the complex vectors

$$
|\psi\rangle=
\begin{bmatrix}
1+i\\
2-i
\end{bmatrix},
\quad
|\phi\rangle=
\begin{bmatrix}
3\\
4i
\end{bmatrix}.
$$

The inner product can be evaluated using `np.vdot(phi, psi)`, corresponding to the bra-ket notation

$$
\langle\phi|\psi\rangle.
$$

Unlike the ordinary dot product, the inner product of complex vectors involves complex conjugation. This ensures that the inner product satisfies the required conjugate symmetry

$$
\langle\phi|\psi\rangle=\langle\psi|\phi\rangle^*.
$$

```python
psi = np.array([1+1j, 2-1j])
phi = np.array([3, 4j])
```

```python
np.vdot(phi, psi)
```

## Norm of a Complex Vector

The inner product of a vector with itself is real and non-negative,

$$
\langle\psi|\psi\rangle=\sum_i |\psi_i|^2.
$$

The corresponding Euclidean norm is therefore

$$
|\psi|=\sqrt{\langle\psi|\psi\rangle}.
$$

This relation provides a convenient way to compute the magnitude of a complex vector using its inner product.

```python
np.sqrt(np.vdot(psi, psi))
```

## Magnitude of the Inner Product

The magnitude of the inner product measures the overlap between two vectors. For the vectors $|\phi\rangle$ and $|\psi\rangle$, the squared magnitude is

$$
\left|\langle\phi|\psi\rangle\right|^2.
$$

This quantity is particularly important in quantum mechanics, where it represents the transition probability between normalized states.

```python
np.abs(np.vdot(phi, psi))**2
```

## Inner Product of Matrices

The inner product can also be extended to matrices by treating the conjugate transpose of one matrix as the corresponding bra. For matrices $\Phi$ and $\Psi$,

$$
\Phi^\dagger\Psi=\left(\Phi^*\right)^T\Psi,
$$

where $\dagger$ denotes the Hermitian conjugate. Consider

$$
\Psi=
\begin{bmatrix}
1+i & 0\\
0 & i
\end{bmatrix},
\quad
\Phi=
\begin{bmatrix}
1 & i\\
1 & 0
\end{bmatrix}.
$$

The Hermitian conjugate of $\Phi$ is

$$
\Phi^\dagger=\left(\Phi^*\right)^T.
$$

Thus, the matrix inner product can be evaluated as

$$
\Phi^\dagger\Psi,
$$

which corresponds directly to the NumPy operation `np.conjugate(Phi.T) @ Psi`.

The conjugate transpose is essential when working with complex-valued vectors and matrices, as it generalizes the ordinary transpose used for real-valued quantities.

```python
Psi = np.array([[1+1j, 0],
                [0, 1j]])
Phi = np.array([[1, 1j],
                [1, 0]])
```

```python
np.conjugate(Phi.T) @ Psi
```
