# Symbolic Computation for Vector Calculus

```python
import sympy as sp
```

## Symbolic Setup

```python
t = sp.symbols("t", real=True)
x = sp.Function("x")(t)
y = sp.Function("y")(t)
z = sp.Function("z")(t)
```

## Particle Kinematics

In classical mechanics, the motion of a particle in three-dimensional space can be described by its **position vector**:

$$
\mathbf{r}(t) =
\begin{bmatrix}
x(t) \\
y(t) \\
z(t)
\end{bmatrix}
$$

```python
r = sp.Matrix([x, y, z])
r
```

From this representation, we can systematically derive higher-order kinematic quantities:

* **Velocity**:
  $$\mathbf{v}(t) = \frac{d\mathbf{r}}{dt}$$

* **Acceleration**:
  $$\mathbf{a}(t) = \frac{d\mathbf{v}}{dt} = \frac{d^2 \mathbf{r}}{dt^2}$$

* **Jerk** (rate of change of acceleration):
  $$\mathbf{j}(t) = \frac{d\mathbf{a}}{dt} = \frac{d^3 \mathbf{r}}{dt^3}$$

```python
v = sp.diff(r, t)
v
```

```python
a = sp.diff(v, t)
a
```

```python
j = sp.diff(a, t)
j
```

Additionally, the magnitudes of vector are defined using the **dot product**:

* **Distance**:
  $$|\mathbf{r}| = \sqrt{\mathbf{r} \cdot \mathbf{r}}$$

* **Speed**:
  $$|\mathbf{v}| = \sqrt{\mathbf{v} \cdot \mathbf{v}}$$

* **Acceleration magnitude**:
  $$|\mathbf{a}| = \sqrt{\mathbf{a} \cdot \mathbf{a}}$$

* **Jerk magnitude**:
  $$|\mathbf{j}| = \sqrt{\mathbf{j} \cdot \mathbf{j}}$$

```python
r_mag = sp.sqrt(r.dot(r))
r_mag
```

```python
v_mag = sp.sqrt(v.dot(v))
v_mag
```

```python
a_mag = sp.sqrt(a.dot(a))
a_mag
```

```python
j_mag = sp.sqrt(j.dot(j))
j_mag
```

As a concrete example, consider a particle undergoing **helical motion**, defined by:

$$
\mathbf{r}(t) =
\begin{bmatrix}
R \cos(\omega t) \\
R \sin(\omega t) \\
v_z t
\end{bmatrix}
$$

where:

* $R$ is the radius of the circular motion
* $\omega$ is the angular velocity
* $v_z$ is the constant velocity along the $z$-axis

This motion combines:

* circular motion in the $xy$-plane
* uniform linear motion along the $z$-axis

```python
R, omega, vz = sp.symbols("R omega v_z", real=True)
```

```python
r_expr = sp.Matrix([
    R * sp.cos(omega * t),
    R * sp.sin(omega * t),
    vz * t
])

r_expr
```

```python
v_expr = sp.diff(r_expr, t)
v_expr
```

```python
a_expr = sp.diff(v_expr, t)
a_expr
```

```python
j_expr = sp.diff(a_expr, t)
j_expr
```

```python
r_mag = sp.sqrt(r_expr.dot(r_expr))
r_mag
```

```python
v_mag = sp.sqrt(v_expr.dot(v_expr))
v_mag
```

```python
a_mag = sp.sqrt(a_expr.dot(a_expr))
a_mag
```

```python
j_mag = sp.sqrt(j_expr.dot(j_expr))
j_mag
```

## Vector Calculus

Define the symbolic variables used throughout this section. Here, $x$, $y$, and $z$ represent Cartesian coordinates, while $t$ is used as the time or path parameter.

```python
x, y, z, t = sp.symbols("x y z t", real=True)
```

The dot product computes a scalar quantity from two vectors. In this example, the work-like quantity is computed using

$$W=\mathbf{F}\cdot\mathbf{r}$$

```python
F = sp.Matrix([2*x, 3*y, z])
r = sp.Matrix([x, y, 4*z])
F.dot(r)
```

The cross product produces a vector perpendicular to both operands. In mechanics, torque is defined as

$$\boldsymbol{\tau}=\mathbf{r}\times\mathbf{F}$$

```python
r = sp.Matrix([x, y, z])
F = sp.Matrix([0, 0, 1])
r.cross(F)
```

The gradient transforms a scalar field into a vector field pointing in the direction of the greatest rate of increase. For the scalar potential $V$, the conservative force is obtained from

$$\mathbf{F}=-\nabla V$$

```python
V = x**2 + y**2 + z**2
-sp.Matrix([sp.diff(V, x), sp.diff(V, y), sp.diff(V, z)])
```

Divergence measures the net outward flux of a vector field from an infinitesimal volume, indicating the presence of sources or sinks. For $\mathbf{F}$ the divergence is

$$\nabla\cdot\mathbf{F}$$

```python
F = sp.Matrix([x**3, 2*y**4, z**2])
sp.diff(F[0], x) + sp.diff(F[1], y) + sp.diff(F[2], z)
```

Curl measures the local rotational tendency of a vector field and is widely used in fluid dynamics and electromagnetism. For $\mathbf{F}$ the curl is

$$\nabla\times\mathbf{F}$$

```python
F = sp.Matrix([y, z, x])
sp.Matrix([
    sp.diff(F[2], y) - sp.diff(F[1], z),
    sp.diff(F[0], z) - sp.diff(F[2], x),
    sp.diff(F[1], x) - sp.diff(F[0], y)
])
```

## Line Integral

A line integral accumulates the contribution of a vector field along a specified path. In this example, the work is evaluated using

$$
W=\int_C\mathbf{F}\cdot d\mathbf{r}=\int_{t_0}^t\mathbf{F}(\mathbf{r}(t))\cdot\frac{d\mathbf{r}}{dt}\,dt.
$$


```python
F = sp.Matrix([x, y])
r = sp.Matrix([t, t])
sp.integrate(F.subs({x: r[0], y: r[1]}).dot(sp.diff(r, t)), (t, 0, 2))
```

The following example evaluates the same vector field along a different path to illustrate how the value of a line integral can depend on the chosen trajectory.

```python
r = sp.Matrix([t**3, t**2])
sp.integrate(F.subs({x: r[0], y: r[1]}).dot(sp.diff(r, t)), (t, 0, 2))
```

## Key Insights

* Dot and cross products describe fundamental vector operations used to compute scalar and vector quantities, respectively.
* Gradient, divergence, and curl are the primary differential operators for analyzing scalar and vector fields.
* Line integrals accumulate the contribution of a vector field along a specified path and provide a mathematical framework for computing quantities such as work.
* Symbolic computation enables vector operations and differential operators to be evaluated exactly, simplifying analytical derivations and verification.
