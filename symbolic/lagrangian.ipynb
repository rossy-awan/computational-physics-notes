# Symbolic Computation for Lagrangian Mechanics

In analytical mechanics, the dynamics of a system are derived from the **Lagrangian**:

$$\mathcal{L}(q, \dot{q}, t) = T - V$$

where:
- $T$ is the kinetic energy
- $V$ is the potential energy
- $q$ represents generalized coordinates

The equations of motion are obtained using the **Euler–Lagrange equation**:

$$\frac{d}{dt} \left( \frac{\partial \mathcal{L}}{\partial \dot{q}} \right) - \frac{\partial \mathcal{L}}{\partial q} = 0$$

In this notebook, we use `sympy` to perform these calculations symbolically.

```python
import sympy as sp
```

## Symbolic Setup

```python
t = sp.symbols("t")
q = sp.Function("q")(t)
```

## Example: Simple Harmonic Oscillator

We consider a mass-spring system with:

$$T = \frac{1}{2} m \dot{q}^2, \quad V = \frac{1}{2} k q^2$$

Thus, the Lagrangian is:

$$\mathcal{L} = \frac{1}{2} m \dot{q}^2 - \frac{1}{2} k q^2$$

```python
m, k = sp.symbols("m k")
qdot = sp.diff(q, t)

T = 1/2 * m * qdot**2
V = 1/2 * k * q**2

L = T - V
L
```

## Deriving the Equation of Motion

We now compute the Euler–Lagrange equation symbolically.

```python
dL_dq = sp.diff(L, q)
dL_dqdot = sp.diff(L, qdot)
dt_dL_dqdot = sp.diff(dL_dqdot, t)

EL = dt_dL_dqdot - dL_dq
sp.simplify(EL)
```

## Solving the Equation of Motion

We can solve the resulting differential equation symbolically.

```python
solution = sp.dsolve(EL, q)
solution
```

## Summary

In this notebook, we demonstrated how symbolic computation can:
* Construct Lagrangians from physical definitions
* Automatically derive equations of motion
* Solve differential equations symbolically

This approach is powerful for:
* rapid prototyping of physical models
* reducing algebraic errors
* exploring complex systems
