# Symbolic Computation for Hamiltonian Mechanics

In analytical mechanics, an equivalent formulation of dynamics is given by the **Hamiltonian**:

$$\mathcal{H}(q, p, t) = p \dot{q} - \mathcal{L}(q, \dot{q}, t)$$

where:

* $p = \frac{\partial \mathcal{L}}{\partial \dot{q}}$ is the **generalized momentum**
* $q$ represents generalized coordinates
* $\mathcal{L}$ is the Lagrangian

For many physical systems, the Hamiltonian corresponds to the total energy:

$$\mathcal{H} = T + V$$

The equations of motion are given by **Hamilton’s equations**:

$$
\dot{q} = \frac{\partial \mathcal{H}}{\partial p}, \qquad
\dot{p} = -\frac{\partial \mathcal{H}}{\partial q}
$$

In this notebook, we use `sympy` to compute these quantities symbolically and derive the equations of motion.

```python
import sympy as sp
```

## Symbolic Setup

```python
t = sp.symbols("t")
q = sp.Function("q")(t)
p = sp.Function("p")(t)
```

## Example: Simple Harmonic Oscillator

We consider a mass-spring system with:

$$T = \frac{1}{2} m \dot{q}^2, \quad V = \frac{1}{2} m \omega^2 q^2$$

Thus, the Hamiltonian is:

$$\mathcal{H} = \frac{1}{2} m \dot{q}^2 + \frac{1}{2} m \omega^2 q^2$$

```python
m, omega = sp.symbols("m omega", positive=True)

T = p**2 / (2 * m)
V = 1/2 * m * omega**2 * q**2

H = T + V
H
```

## Deriving the Equation of Motion

We now compute the equation of motion symbolically.

```python
dq_dt = sp.diff(H, p)
dp_dt = -sp.diff(H, q)
eq1 = sp.Eq(sp.diff(q, t), dq_dt)
eq2 = sp.Eq(sp.diff(p, t), dp_dt)
```

```python
eq1
```

```python
eq2
```

## Solving the Equation of Motion

We can solve the resulting differential equation symbolically.

```python
ode = sp.Eq(sp.diff(q, t, 2), -omega**2 * q)
ode
```

```python
solution = sp.dsolve(ode, q)
solution
```

```python
q_sol = solution.rhs
p_sol = m * sp.diff(q_sol, t)
sp.simplify(p_sol)
```

## Summary

In this notebook, we demonstrated how symbolic computation can:
* Construct Hamiltonians from physical definitions
* Automatically derive equations of motion
* Solve differential equations symbolically

This approach is powerful for:
* rapid prototyping of physical models
* reducing algebraic errors
* exploring complex systems
